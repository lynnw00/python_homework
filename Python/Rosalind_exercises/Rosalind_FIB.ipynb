```python
def rabbit_pairs(n, k):
    if n == 1 or n == 2:
        return 1
    
    prev2 = 1  # F(n-2)
    prev1 = 1  # F(n-1)
    
    for month in range(3, n+1):
        current = prev1 + k * prev2
        prev2, prev1 = prev1, current
    
    return prev1
```

```python
n = 35
k = 2
print(rabbit_pairs(n, k))
```
