```python
def gc_content(seq):
    "Calculate the GC content of a sequence (0-100)"
    total_Cs = seq.count('C')
    total_Gs = seq.count('G')
    gc_percent = (total_Cs + total_Gs) / len(seq)
    return 100 * gc_percent
```

```python
file = "C:/Users/Lynnw/Downloads/rosalind_gc.txt"

#file = "C:/Users/Lynnw/Python/Rosalind_exercises/GC_data.txt"

sequences = {}
current_id = None

with open(file, "r") as f:
    for line in f:
        line = line.strip()
        if not line:
            continue
        if line.startswith('>'):
            current_id = line[1:]
            sequences[current_id] = ""
        else:
            sequences[current_id] += line

max_id = max(sequences, key=lambda id: gc_content(sequences[id]))
max_gc = gc_content(sequences[max_id])

print(max_id)
print(f"{max_gc:.6f}")
```
