```python
def is_invalid(s):
    if len(s)%2 == 1:
        return False
    middle = len(s)//2
    first_half = s[:middle]
    second_half = s[middle:]
    if first_half == second_half:
        return True
    else:
        return False
```

need a function that takes a range and returns a list of invalid IDs

a range = "11-22"
invalid IDs = 11, 22

```python
r = '11-22'
start, stop = r.split('-')
start = int(start)
stop = int(stop)
results = []

for number in range(start, stop +1):
    validity = is_invalid(str(number))
    #print(number, validity)

    if validity == True:
        results.append(number)

print(f'The invalid IDs are {results}')
```
