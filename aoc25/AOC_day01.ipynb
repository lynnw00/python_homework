```python
path = "C:/Users/Lynnw/Python/AOC/"
file_name = "data_day01.txt"

with open(path + file_name, "r") as file:
    data = file.read()
    
words = data.split() 
rotations = []    
for w in words:
    clean_w = w.strip().replace('"', '')
    if clean_w != "":
        rotations.append(clean_w)
```

```python
def rotate_dial(dial, r):
    direction = r[0]
    steps = int(r[1:])

    if direction == 'L':
        dial = dial - steps
    else:
        dial = dial + steps
    
    while dial > 99:
        dial -= 100
    while dial < 0:
        dial += 100
        
    return dial

    # can also use % 100, gives the remaining number.
```

```python
#rotations = ["L68", "L30", "R48", "L5", "R60", "L55", "L1", "L99", "R14", "L82"]
dial = 50
count = 0

for rotation in rotations:
    dial = rotate_dial(dial, rotation)
    #print(f'The dial is rotated {rotation} to point at {dial}')
    if dial == 0:
        count += 1
        
print(f'password = {count}')

file.close()
```
