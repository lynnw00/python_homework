```python
path = "C:/Users/Lynnw/Python/AOC/"
file_name = "data_day01.txt"

with open(path + file_name, "r") as file:
    data = file.read()

words = data.split() 
rotations = []    
for w in words:
    clean_w = w.strip().replace('"', '')
    if clean_w != "":
        rotations.append(clean_w)
```

```python
def part2(instructions: list[str], start=50, dial_nums=100):
    zero_counter = 0
    curr_pos = start

    for instr in instructions:
        direction = instr[0]
        steps = int(instr[1:])

        if direction == "R":
            zero_counter += (curr_pos + steps) // dial_nums
            curr_pos = (curr_pos + steps) % dial_nums

        else:  # L
            new_pos = (curr_pos - steps) % dial_nums

            if curr_pos == 0:
                zero_counter += steps // dial_nums
            else:
                if steps > curr_pos:
                    zero_counter += ((steps - curr_pos - 1) // dial_nums) + 1
                    if new_pos == 0:
                        zero_counter += 1
                elif steps == curr_pos:
                    zero_counter += 1

            curr_pos = new_pos

    return zero_counter

result = part2(rotations)
print("Part 2 result:", result)
```
