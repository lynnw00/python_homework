```python
from util import file_to_list
path = "C:/Users/Lynnw/Python/AOC/data_day06.txt"
```

```python
def ex_hw(path):
    lines = file_to_list(path)

    # last line contains + or *
    last_line = lines.pop()
    operations = last_line.split()

    # count columns of numbers
    first_line_numbers = lines[0].split()
    column_count = len(first_line_numbers)

    # create lists, one per column
    numbers = []
    for i in range(column_count):
        numbers.append([])

    for line in lines:
        values = line.split()

        # add number to correct column
        for i in range(len(values)):
            numbers[i].append(int(values[i]))

    return numbers, operations
```

```python
def product(numbers):
    result = 1
    for num in numbers:
        result = result * num
    return result
```

```python
def do_hw(numbers, operations):
    total = 0

    for i in range(len(numbers)):
        current_numbers = numbers[i]
        current_operation = operations[i]

        if current_operation == '+':
            column_result = sum(current_numbers)
        elif current_operation == '*':
            column_result = product(current_numbers)

        total = total + column_result

    return total
```

```python
# final result
numbers, operations = ex_hw(path)
answer = do_hw(numbers, operations)
print(answer)
```

```python
def ex_hw_new(path):
    lines = file_to_list(path, raw=True)
    
    last_line = lines.pop()
    operations = last_line.split()

    # Number of characters in each line (columns)
    column_count = len(lines[0])

    # Create empty strings for each column
    columns = []
    for i in range(column_count):
        columns.append("")

    # Build each column by reading characters vertically
    for col_index in range(column_count):
        for line in lines:
            columns[col_index] += line[col_index]

    # Convert column strings into lists of numbers
    numbers = []
    current_group = []

    for column in columns:
        no_spaces = column.replace(" ", "")

        if no_spaces != "":
            current_group.append(int(no_spaces))
        else:
            numbers.append(current_group)
            current_group = []
    numbers.append(current_group)

    return numbers, operations
```

```python
numbers, operations = ex_hw_new(path)
do_hw(numbers, operations)
```
