Per bank, go from left to right digits, first time going over it, find the highest number, then go over it again starting at that number, to find the next highest number.
\
If the 9 is at the last position, this will not work.
\
Sum the output numbers.

```python
path = "C:/Users/Lynnw/Python/AOC/"
file_name = "data_day03.txt"

with open(path + file_name, "r") as file:
    data = file.read()
    
words = data.split() 
jolts = []    
for w in words:
    clean_w = w.strip().replace('"', '')
    if clean_w != "":
        jolts.append(clean_w)
```

```python
s = "811111111111119"

i = 0
print(max(s[i+1:]))
```

```python
def maximum_digit(s):
    max_digit = 0
    max_index = -1
    for i, n in enumerate(s):
        if int(n) > max_digit:
            max_digit = int(n)
            max_index = i
    
    return (max_digit, max_index)
```

```python
s = '818181911112111'
first, i = maximum_digit(s[:-1])
second, j = maximum_digit(s[i+1:])
```

```python
bank = jolts
final = 0
added = 0
for jolt in bank:
    first, i = maximum_digit(jolt[:-1])
    second, j = maximum_digit(jolt[i+1:])
    final = str(first) + str(second)
    added += int(final)
print(added)
```
