```python
def adjacent(pos):
    x, y = pos
    return [
        (x-1, y-1), (x-1, y), (x-1, y+1),
        (x,   y-1),          (x,   y+1),
        (x+1, y-1), (x+1, y), (x+1, y+1)
    ]    
```

# Part two
For every roll that can be removed, change the '@' to a '.'

Add a counter for the amount of rolls removed.

```python
grid = [list(row) for row in data]
# Changed each row into a list of characters, needed to be able to change it.

height = len(grid)
occupied = '@'
removed_total = 0

while True:
    to_remove = []

    # check every cell
    for y, row in enumerate(grid):
        for x, value in enumerate(row):
            if value != occupied:
                continue

            # get valid neighbors
            neighbors = adjacent((x, y))
            valid_neighbors = [
                (nx, ny) for nx, ny in neighbors
                if 0 <= ny < height and 0 <= nx < len(grid[ny])
            ]

            # count occupied neighbors
            occupied_count = sum(grid[ny][nx] == occupied for nx, ny in valid_neighbors)

            # mark for removal if fewer than 4 neighbors
            if occupied_count < 4:
                to_remove.append((x, y))
                # do it simultaneously so it does not error

    # stop if nothing to remove
    if not to_remove: #if length list = 0, then break
        break

    # remove all marked rolls and update it!
    for x, y in to_remove:
        grid[y][x] = '.'

    removed_total += len(to_remove)

print("Total removed rolls:", removed_total)
```
